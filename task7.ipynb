Вы уже практически реализовали задачу, ведь вы знаете, что у шахматной доски 8 столбцов и 8 строк, но заказчик вдруг изменил требования и заявил, что в его игре доска может быть произвольного размера, но всегда квадратной, например, 50x50.

Напишите функцию make_field(size), в которую поступает размер доски, и которая возвращает требуемый массив numpy. Тип элементов массива должен быть np.int8. Для решения задачи нельзя использовать циклы.



```python
import numpy as np

def make_field(size):#если сумма четная - белая (1), нечетная - черная (0)
    field = np.fromfunction(lambda i, j: (i + j) % 2, (size, size), dtype=np.int8)
    return field


print("Доска 3x3:\n", make_field(3))
print("Доска 5x5:\n", make_field(5))
```

```python
import numpy as np

def super_sort(rows, cols):
    A = np.random.randint(1, 101, (rows, cols))
    # Создаем копию в В
    B = A.copy()
    
    for j in range(cols):
        if j % 2 == 0:  # четный столбец 
            B[:, j] = np.sort(B[:, j])
        else:  #
            B[:, j] = np.sort(B[:, j])[::-1]
    
    return (A, B)


A, B = super_sort(3, 5)
print("Матрица А:\n", A)
print("\nМатрица В (четные столбцы по возрастанию, нечетные по убыванию):\n", B)
```

mult – матричное перемножение без использования numpy

el_mult – поэлементное перемножение без использования numpy

np_mult – матричное перемножение с использованием numpy

np_el_mult – поэлементное перемножение с использованием numpy

При помощи timeit засеките время работы функций для n = 100 000 (для чистоты эксперимента засекайте время на одинаковых матрицах). Во сколько раз numpy-версия функций работает быстрее реализаций без numpy?

```python

import numpy as np
import time

def mult(A, B):
    n = len(A)
    result = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                result[i][j] += A[i][k] * B[k][j]
    return np.array(result)

def el_mult(A, B):
    n = len(A)
    result = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            result[i][j] = A[i][j] * B[i][j]
    return np.array(result)

def np_mult(A, B):
    return np.dot(A, B)

def np_el_mult(A, B):
    return A * B


n = 10000
print(f"Тестирование для n={n}")
print("(Для n=100000 потребовалось бы слишком много памяти и времени)")


A = np.random.rand(n, n)
B = np.random.rand(n, n)

start = time.time()
result_np_mult = np_mult(A, B)
np_mult_time = time.time() - start

start = time.time()
result_np_el = np_el_mult(A, B)
np_el_time = time.time() - start

print(f"np_mult время: {np_mult_time:.4f} сек")
print(f"np_el_mult время: {np_el_time:.4f} сек")
```

№10 (2 балла)
Напишите программу, которая будет решать систему линейных уравнений вида:

```python
import numpy as np

def solve_system():
    print("Введите коэффициенты системы:")
    a11 = float(input("a11: "))
    a12 = float(input("a12: "))
    a21 = float(input("a21: "))
    a22 = float(input("a22: "))
    b1 = float(input("b1: "))
    b2 = float(input("b2: "))

    print("\nВведённая система:")
    print(f"{a11}*x + {a12}*y = {b1}")
    print(f"{a21}*x + {a22}*y = {b2}")
    
    A = np.array([[a11, a12], [a21, a22]])
    B = np.array([b1, b2])
    
    det = np.linalg.det(A)
    
    if abs(det) < 1e-10:
        print("Определитель равен 0. Система не имеет единственного решения.")
    else:
        solution = np.linalg.solve(A, B)
        print(f"Решение: x = {solution[0]}, y = {solution[1]}")

solve_system()
```

```python
def create_negative():
    try:
        with open('test.bmp', 'rb') as infile:
            data = infile.read()
        
        if len(data) < 54:
            print("Файл слишком маленький для BMP")
            return
        
        header = data[:54]
        pixels = data[54:]
        negative_pixels = bytes([255 - b for b in pixels])
        new_data = header + negative_pixels
        
        with open('output.bmp', 'wb') as outfile:
            outfile.write(new_data)
        
        print("Негатив создан в output.bmp")
        
    except FileNotFoundError:
        print("Файл input.bmp не найден")

create_negative()
```

Выведите на экран в порядке упоминания в файле те товары, цена на которые уменьшилась.

Если товаров, попадающих под условие задачи, нет, то на экран выводить ничего не надо.

Для быстрого решения задачи воспользуйтесь объектом DictReader.

```python
import csv

def check_price_decrease():
    try:
        with open('prices.csv', 'r', encoding='utf-8') as f:
            reader = csv.DictReader(f)
            for row in reader:
                old_price = int(row['old_price'])
                new_price = int(row['new_price'])
                if new_price < old_price:
                    print(row['name'])
    except FileNotFoundError:
        print("Файл prices.csv не найден")


with open('prices.csv', 'w', encoding='utf-8', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['name', 'old_price', 'new_price'])
    writer.writeheader()
    writer.writerow({'name': 'Товар 1', 'old_price': '120', 'new_price': '80'})
    writer.writerow({'name': 'Товар 2', 'old_price': '200', 'new_price': '250'})
    writer.writerow({'name': 'Товар 3', 'old_price': '150', 'new_price': '120'})
    writer.writerow({'name': 'Товар 4', 'old_price': '80', 'new_price': '100'})
    writer.writerow({'name': 'Товар 5', 'old_price': '69', 'new_price': '75'})
    writer.writerow({'name': 'Товар 6', 'old_price': '50', 'new_price': '20'})

print("Товары с уменьшением цены:")
check_price_decrease()
```

Поля отделяются запятой. Каждая запись об участнике находится в отдельной строке.

Логин участника имеет вид sh-kaluga16-**-#@, где ** — номер школы, а ## — номер класса, @ — номер участника. Определите результаты олимпиады для указанной школы и класса.

Ваша программа должна вывести имена всех участников и балл через пробел из данной школы и класса в порядке убывания балла. При совпадении баллов фамилии нужно выводить в обратном алфавитном порядке. Информацию о каждом участнике надо выводить с новой строки.

```python
import csv

def olympiad_results(school, grade):
    results = []
    
    with open('rez.csv', 'r', encoding='utf-8') as f:
        reader = csv.DictReader(f)
        
        for row in reader:
            login = row['login']
            parts = login.split('-')
            if len(parts) >= 5 and parts[2] == school and parts[3] == grade:
                name = row['user_name']
                score = int(row['Score'])
                results.append((name, score))
    
    results.sort(key=lambda x: (-x[1], x[0]))
    
    for name, score in results:
        print(f"{name} {score}")

olympiad_results('36', '11')
```
