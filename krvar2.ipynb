Задание 2 Дан одномерный массив числовых значений, насчитывающий N элементов.
Выполнить перемещение элементов массива по кругу вправо, т. е. A[1] → A[2]; A[2] → A[3];
... A[n] → A[1]

```python
def task2(arr):
    if len(arr) <= 1:
        return arr
    return [arr[-1]] + arr[:-1]


arr = [1, 2, 3, 4, 5]
print(f"Исходный массив: {arr}")
print(f"После сдвига вправо: {task2(arr)}")
```

Задание 3 (1 балл) Заданы M строк слов, которые вводятся с клавиатуры (в каждой строке – одно
слово). Вводится слог (последовательность букв). Подсчитать количество таких слогов в
каждой строке

```python
def task3():
    m = int(input("Введите количество строк: "))
    results = []
    for i in range(m):
        word = input(f"Строка {i+1} (одно слово): ")
        syllable = input("Введите слог для поиска: ")
        count = word.count(syllable)
        results.append(count)
        print(f"В слове '{word}' слог '{syllable}' встречается {count} раз(а)")
    return results

task3()
```

Задание 5  (2 балла) Напишите функцию, принимающую от 1 до 3 параметров — целых чисел (как
стандартная функция range). Функция должна выдавать список кубов чисел. Если подан 1
аргумент — от 0 до этого числа (не включительно). Если поданы 2 аргумента, то первый
интерпретируется как начальное число, второй — как конечное (не включительно). Если
поданы 3 аргумента, то третий аргумент интерпретируется как шаг. 


```python
def range_cubes(*args):
    if len(args) == 1:
        start, stop, step = 0, args[0], 1
    elif len(args) == 2:
        start, stop, step = args[0], args[1], 1
    elif len(args) == 3:
        start, stop, step = args[0], args[1], args[2]
    else:
        return []
    
    return [i**3 for i in range(start, stop, step)]

print(f"range_cubes(5): {range_cubes(5)}")          
print(f"range_cubes(2, 6): {range_cubes(2, 6)}")     
print(f"range_cubes(1, 10, 2): {range_cubes(1, 7, 2)}")
```

Задание 6 6. (2 балла) Дана строчка текста, состоящая из слов и пробелов. Словом считается
последовательность символов, слова разделены одним или большим числом пробелов.
Для каждого слова этого текста узнайте порядковый номер его вхождения в текст именно в
той форме, в которой указано слово. Для первого вхождения слова выведите «1», для
второго вхождения того же слова выведите «2» и так далее. 

```python
def task6(text):
    words = text.split()
    word_counts = {}
    result = []
    for word in words:
        word_counts[word] = word_counts.get(word, 0) + 1
        result.append(str(word_counts[word]))
    return " ".join(result)


text = "hello world hello vsem hello world"
print(f"Текст: '{text}'")
print(f"Результат: {task6(text)}")
```

Задание 7  7. (2 балла) Вывести результат нахождения квадратного корня всех элементов списка, которые
больше 55. Предоставить два решения: 1) при помощи функций map и filter, 2) при помощи
списочного выражения. 

```python
def task7(nums):
    #map и filter
    res1 = list(map(lambda x: x**0.5, filter(lambda x: x > 55, nums)))
    #списковое выражение
    res2 = [x**0.5 for x in nums if x > 55]
    return res1, res2

numbers = [10, 64, 100, 36, 81, 49, 60]
print(f"Исходный список: {numbers}")
res1, res2 = task7(numbers)
print(f"Способ 1 (map+filter): {res1}")
print(f"Способ 2 (list comp): {res2}")
```

Задание 8 8. (2 балла) Напишите функцию count_cond(condition), которая возвращает функцию одного
аргумента n. Эта функция считает, сколько из чисел от 1 до n включительно удовлетворяют
условию condition - предикату от двух аргументов, где первый - это число n, а второй - число
от 1 до n включительно. 


```python
def count_cond(condition):
    def counter(n):
        count = 0
        for i in range(1, n + 1):
            if condition(n, i):
                count += 1
        return count
    return counter


count_factors = count_cond(lambda n, i: n % i == 0)
print(f"count_factors(2): {count_factors(2)}")   
print(f"count_factors(4): {count_factors(4)}")   
print(f"count_factors(12): {count_factors(10)}") 

```

Задание 9 9. (2 балла) Напишите функцию compose1(f, g), которая возвращает суперпозицию функций
f(g(x)) от x. 


```python
def compose1(f, g):
    return lambda x: f(g(x))

add_one = lambda x: x + 1
square = lambda x: x**2
a1 = compose1(square, add_one)
print(f"a1(4) = ((4+1)^2) = {a1(4)}")  

mul_three = lambda x: x * 3
a2 = compose1(mul_three, a1)
print(f"a2(6) = ((6+1)^2)*3 = {a2(6)}")  
print(f"a2(7) = ((7+1)^2)*3 = {a2(7)}")
```

Задание 10  10. (2 балла) Реализуйте иерархию классов:
 Автомобиль (базовый класс)
 Легковой
 Грузовой
 Пассажирский
Поля определите самостоятельно. Должен быть доступен расчёт расхода топлива в
зависимости от загрузки и расчёт стоимости и времени поездки. Продемонстрируйте работу
классов. 


```python
class Car:
    def __init__(self, fuel_per_100km, speed_kmh):
        self.fuel_per_100km = fuel_per_100km  
        self.speed_kmh = speed_kmh            
    
    def fuel_for_distance(self, km, load_kg=0):
        extra = (load_kg // 100) * 0.01 * self.fuel_per_100km
        return (self.fuel_per_100km + extra) * km / 100
    
    def time_for_distance(self, km):
        return km / self.speed_kmh
    
    def cost_for_distance(self, km, fuel_price_per_liter=50, load_kg=0):
        return self.fuel_for_distance(km, load_kg) * fuel_price_per_liter

class PassengerCar(Car):
    def __init__(self, fuel_per_100km, speed_kmh, seats):
        super().__init__(fuel_per_100km, speed_kmh)
        self.seats = seats
        self.type = "Легковой"

class Truck(Car):
    def __init__(self, fuel_per_100km, speed_kmh, max_load_kg):
        super().__init__(fuel_per_100km, speed_kmh)
        self.max_load_kg = max_load_kg
        self.type = "Грузовой"

class Bus(Car):
    def __init__(self, fuel_per_100km, speed_kmh, passenger_count):
        super().__init__(fuel_per_100km, speed_kmh)
        self.passenger_count = passenger_count
        self.type = "Пассажирский"


passenger = PassengerCar(8, 120, 5)
truck = Truck(25, 80, 25000)
bus = Bus(15, 90, 50)

distance = 500

print(f"{passenger.type} автомобиль:")
print(f"  Расход топлива на {distance} км: {passenger.fuel_for_distance(distance):.1f} л")
print(f"  Время в пути: {passenger.time_for_distance(distance):.1f} ч")
print(f"  Стоимость поездки: {passenger.cost_for_distance(distance):.0f} руб")

print(f"\n{truck.type} автомобиль (с грузом 2000 кг):")
print(f"  Расход топлива на {distance} км: {truck.fuel_for_distance(distance, 2000):.1f} л")
print(f"  Время в пути: {truck.time_for_distance(distance):.1f} ч")

print(f"\n{bus.type} автобус (с {bus.passenger_count} пассажирами):")
print(f"  Расход топлива на {distance} км: {bus.fuel_for_distance(distance):.1f} л")
print(f"  Время в пути: {bus.time_for_distance(distance):.1f} ч")
```

Задание 11 11. (4 балла) Напишите класс WheelBeing (Колесник). Экземпляр класса при инициализации
принимает аргументы: количество насекомых, вид насекомых, объем колесника.
Класс обеспечивает выполнение методов (wb – экземпляр класса):
 экземпляры класса можно сравнивать: сначала по объему, затем по количеству
насекомых, затем по виду по алфавиту;
 из экземпляра класса можно вычесть число: wb -= number, при этом меняются
значения атрибутов:
из количества насекомых вычитается число, если остается значение не меньше 10,
иначе просто 10;
вид не меняется;
объем – половина от исходного (при нечетном числе – меньшая часть);
 к экземпляру можно прибавить другой экземпляр: wb2 = wb + wb1, возвращается
новый экземпляр с атрибутами:
количество насекомых и объемы равны средним значениям, округленным до
целого вниз;
вид – меньший по алфавиту;
исходные экземпляры не меняются;
 экземпляр класса можно разделить на число: res = wb / number, возвращается
список экземпляров класса длиной, равной числу, с атрибутами:
вид – исходный с добавлением числа 1, 2 и т. д.
количество – целая часть от деления исходного количества на число;
объем – такой же, как у исходного;
 экземпляр класса можно вызвать с аргументом-числом – возвращается
значение количество * число // объем;
 add_insects() – добавить насекомых – вызывается с аргументом-числом, при
положительном значении добавляет к исходному количеству, при отрицательном
– уменьшает на переданное значение;
 __str__ – возвращает строку:
Wheel being with a volume of <объем> accommodates <количество насекомых>
insects of the species <вид>.
 __repr__ – возвращает строку:
WheelBeing(<количество насекомых>, '<вид>', <объем>)

```python
class WheelBeing:
    def __init__(self, insects, species, volume):
        self.insects = insects
        self.species = species
        self.volume = volume
    
    # Сравнение
    def __gt__(self, other):
        if self.volume != other.volume:
            return self.volume > other.volume
        if self.insects != other.insects:
            return self.insects > other.insects
        return self.species > other.species
    
    def __lt__(self, other):
        if self.volume != other.volume:
            return self.volume < other.volume
        if self.insects != other.insects:
            return self.insects < other.insects
        return self.species < other.species
    
    def __eq__(self, other):
        return self.volume == other.volume and self.insects == other.insects and self.species == other.species
    
    def __ne__(self, other):
        return not self.__eq__(other)
    
    def __ge__(self, other):
        return self > other or self == other
    
    def __le__(self, other):
        return self < other or self == other
    
    # wb -= number
    def __isub__(self, number):
        self.insects -= number
        if self.insects < 10:
            self.insects = 10
        self.volume = self.volume // 2
        return self
    
    # wb + wb1
    def __add__(self, other):
        return WheelBeing(
            (self.insects + other.insects) // 2,
            min(self.species, other.species),
            (self.volume + other.volume) // 2
        )
    
    # wb / number
    def __truediv__(self, number):
        result = []
        for i in range(number):
            result.append(WheelBeing(self.insects // number, f"{self.species}{i+1}", self.volume))
        return result
    
    # wb(число)
    def __call__(self, number):
        return self.insects * number // self.volume
    
    def add_insects(self, number):
        self.insects += number
    
    def __str__(self):
        return f"Wheel being with a volume of {self.volume} accommodates {self.insects} insects of the species {self.species}."
    
    def __repr__(self):
        return f"WheelBeing({self.insects}, '{self.species}', {self.volume})"




#1
wb = WheelBeing(120, "spider", 15)
wbl = WheelBeing(120, "fly", 15)
print(wb > wbl, wb != wbl, wb < wbl)
print(wb)
print(wbl)
print()

wb -= 117
wbl.add_insects(-35)
wb2 = wb + wbl
print(wb)
print(wbl)
print(wb2)
print()

#2
wb3 = WheelBeing(500, "gnat", 8)
wbl3 = WheelBeing(498, "midge", 3)
print(wb3 >= wbl3, wb3 == wbl3, wb3 < wbl3)
print(wb3)
print(wbl3)
print()

wb4 = wb3 + wbl3
print(wb4)
print(wb4(3))
print(wbl3(17))
print()

res = wb3 / 3
res[0] -= 10
print(res)
print(wb3)
```
