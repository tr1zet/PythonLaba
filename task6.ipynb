Напишите класс Balance для описания весов с двумя чашами. На левую и правую чашу объекта будут добавляться грузы с различным весом, ваша задача определить положение чаш.

```python
class Balance:
    def __init__(self):
        self.left_weight = 0
        self.right_weight = 0
    
    def add_right(self, weight):
        self.right_weight += weight
    
    def add_left(self, weight):
        self.left_weight += weight
    
    def result(self):
        if self.left_weight == self.right_weight:
            return "="
        elif self.left_weight > self.right_weight:
            return "L"
        else:
            return "R"



balance = Balance()
balance.add_right(10)
balance.add_left(5)
balance.add_left(5)
print(balance.result())
balance.add_left(1)
print(balance.result())

```

### №3 (1 балл)

Напишите класс Selector. Экземпляр этого класса при инициализации получает список чисел. Вызов метода get_odds возвращает нечётные числа из первоначального списка, вызов get_evens — чётные. Числа должны идти в том же порядке, в котором они были в изначальном списке.


```python
class Selector:
    def __init__(self, numbers):
        self.numbers = numbers
    
    def get_odds(self):
        return [n for n in self.numbers if n % 2 != 0]
    
    def get_evens(self):
        return [n for n in self.numbers if n % 2 == 0]


selector = Selector([11, 12, 13, 14, 15, 16, 17, 8, 19, 10])
print("Нечётные:", selector.get_odds())
print("Чётные:", selector.get_evens())

```

Определите класс Point. При инициализации экземпляру передаются координаты х и у.

При сравнении двух экземпляров оператор == должен возвращать True, если координаты точек равны, и False — если нет. При сравнении оператором != должно возвращаться True, если координаты точек не равны, и False — если равны.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    def __ne__(self, other):
        return not self.__eq__(other)



p1 = Point(1, 2)
p2 = Point(5, 6)
p3 = Point(3, 4)
p4 = Point(1, 2)

print(p1 == p2)
print(p1 != p2)

print(p1 == p3)
print(p1 != p3)

print(p1 == p4)
print(p1 != p4)
```

- r1 = ReversedList(list) — создание обратного списка.

- r[0] — доступ к элементам в обратном порядке. r[0] — последний элемент первоначального списка. r[1] — предпоследний и так далее.

```python
class ReversedList:
    def __init__(self, lst):
        self.lst = lst
    
    def __len__(self):
        return len(self.lst)
    
    def __getitem__(self, index):
        return self.lst[len(self.lst) - 1 - index]


r1 = ReversedList([10, 20, 30])
for i in range(len(r1)):
    print(r1[i])

print()

r2 = ReversedList([])
print(len(r2))
print()

r3 = ReversedList([10])
print(len(r3))
print(r3[0])
```

- arr = SparseArray() — создание пустого разреженного массива.
- arr[i] = value — записать значение в массив.
- arr[i] — прочитать значение из массива.

```python
class SparseArray:
    def __init__(self):
        self.data = {}
    
    def __setitem__(self, index, value):
        if value != 0:
            self.data[index] = value
        elif index in self.data:
            del self.data[index]
    
    def __getitem__(self, index):
        return self.data.get(index, 0)


# Демонстрация работы
arr = SparseArray()
arr[1] = 10
arr[2] = 20
arr[8] = 20

for i in range(10):
    print(f"arr[{i}] = {arr[i]}")

print()

arr2 = SparseArray()
arr2[1000000000] = 123
print(f"arr2[999999999] = {arr2[999999999]}")
print(f"arr2[1000000000] = {arr2[1000000000]}")
print(f"arr2[1000000001] = {arr2[1000000001]}")



```

Реализуйте класс Polynomial для вычисления значений многочленов при заданных x, а также для сложения многочленов.

poly = Polynomial(coefficients) — создать новый многочлен с коэффициентами coefficients.  
coefficients[0] — свободный член, coefficients[1] — множитель при \(x^1\), coefficients[2] — множитель при \(x^2\), и так далее.

y = poly(x) — вычислить значение многочлена в точке x.

\[y = c_0 + c_1 x^1 + c_2 x^2 + \dots + c_n x^n\]

poly_sum = poly1 + poly2 — результат сложения двух многочленов должен быть экземпляром Polynomial и содержать коэффициенты, которые получаются в результате сложения коэффициентов poly1 и poly2 при одинаковых степенях x.

**Примечание:** модуль numpy в этой задаче использовать нельзя.




```python
class Polynomial:
    def __init__(self, coefficients):
        self.coefficients = coefficients
    
    def __call__(self, x):
        result = 0
        for power, coeff in enumerate(self.coefficients):
            result += coeff * (x ** power)
        return result
    
    def __add__(self, other):
        max_len = max(len(self.coefficients), len(other.coefficients))
        new_coeffs = [0] * max_len
        for i in range(len(self.coefficients)):
            new_coeffs[i] += self.coefficients[i]
        for i in range(len(other.coefficients)):
            new_coeffs[i] += other.coefficients[i]
        return Polynomial(new_coeffs)
    
    def __str__(self):
        return f"Polynomial({self.coefficients})"



poly = Polynomial([10, -1])
print(poly(0))
print(poly(1))
print(poly(2))

print()

poly1 = Polynomial([0, 0, 1])
print(poly1(-2))
print(poly1(-1))
print(poly1(0))
print(poly1(1))
print(poly1(2))

poly2 = Polynomial([0, 0, 2])
print(poly2(-2))
print(poly2(-1))
print(poly2(0))
print(poly2(1))
print(poly2(2))
print()

poly3 = poly1 + poly2
print(poly3(-2))
print(poly3(-1))
print(poly3(0))
print(poly3(1))
print(poly3(2))
print()
```

### №8 (1 балл)

Реализуйте класс Queue, моделирующий работу очереди (принцип FIFO). Очередь может быть инициализирована различным числом элементов.

q = Queue(1, 2, 3, 4, 5) – создаёт очередь [1 -> 2 -> 3 -> 4 -> 5]

Класс Queue реализует следующие методы:

- append(values) – добавляет несколько значений в конец очереди (как минимум одно).
- copy() – создаёт копию данной очереди.
- pop() – вытаскивает и возвращает первый элемент очереди, при этом этот элемент из очереди удаляется. Если очередь пуста, то возвращает None.
- extend(queue) – расширяет данную очередь другой.
- next() – возвращает новую очередь, начинающуюся со второго элемента текущей.

Операторы и встроенные функции:

- queue1 + queue2 – склейка очередей создаёт новую увеличенную очередь.
- queue1 += queue2 – расширяет первую очередь второй.
- queue1 == queue2 – проверяет очереди на равенство всех элементов.
- queue >> N – создаёт новую очередь без первых N элементов.
- str(queue) – приводит очередь к строке вида [q1 -> q2 -> ... -> qn].
- next(queue) – аналогичное действие методу next().

```python
class Queue:
    def __init__(self, *args):
        self.items = list(args)
    
    def append(self, *values):
        for value in values:
            self.items.append(value)
    
    def copy(self):
        new_queue = Queue()
        new_queue.items = self.items.copy()
        return new_queue
    
    def pop(self):
        if len(self.items) == 0:
            return None
        return self.items.pop(0)
    
    def extend(self, queue):
        self.items.extend(queue.items)
    
    def next(self):
        if len(self.items) <= 1:
            return Queue()
        return Queue(*self.items[1:])
    
    def __len__(self):
        return len(self.items)
    
    def __str__(self):
        if not self.items:
            return "[]"
        return "[" + " -> ".join(str(item) for item in self.items) + "]"
    
    def __add__(self, other):
        new_queue = self.copy()
        new_queue.extend(other)
        return new_queue
    
    def __iadd__(self, other):
        self.extend(other)
        return self
    
    def __eq__(self, other):
        if not isinstance(other, Queue):
            return False
        return self.items == other.items
    
    def __rshift__(self, n):
        if n >= len(self.items):
            return Queue()
        return Queue(*self.items[n:])

    def __iter__(self):
        return self

    def __next__(self):
        if len(self.items) == 0:
            raise StopIteration
        return self.pop()



q1 = Queue(1, 2, 3)
print(q1)

q1.append(4, 5)
print(q1)

qx = q1.copy()
print(qx.pop())
print(qx)

q2 = q1.copy()
print(q2)

print(q1 == q2, id(q1) == id(q2))

q3 = q2.next()
print(q1)
print(q2)
print(q3)

print(q1 + q3)

q3.extend(Queue(1, 2))
print(q3)

q4 = Queue(1, 2)
q4 += q3 >> 4
print(q4)

q5 = next(q4)
print(q4)
print(q5)
```

```python
class Triangle:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    
    def perimeter(self):
        return self.a + self.b + self.c


class EquilateralTriangle(Triangle):
    def __init__(self, side):
        super().__init__(side, side, side)


triangle = Triangle(3, 4, 5)
print(f"Периметр треугольника: {triangle.perimeter()}")

equilateral = EquilateralTriangle(6)
print(f"Периметр равностороннего треугольника: {equilateral.perimeter()}")

```

Реализуйте несколько классов, вычисляющих суммы последовательностей от 1 до N: обычную, квадратов и кубов. Базовый класс Summarator реализует методы transform(n) (тождественное преобразование) и sum(N) (сумма transform(n) от 1 до N). Производные: SquareSummarator (transform(n) = n**2), CubeSummarator (transform(n) = n**3).

```python
class Summarator:
    def transform(self, n):
        return n
    
    def sum(self, N):
        total = 0
        for n in range(1, N + 1):
            total += self.transform(n)
        return total


class SquareSummarator(Summarator):
    def transform(self, n):
        return n ** 2


class CubeSummarator(Summarator):
    def transform(self, n):
        return n ** 3



N = 5

summator = Summarator()
print(f"Сумма 1..{N}: {summator.sum(N)}")
square_summator = SquareSummarator()
print(f"Сумма квадратов 1..{N}: {square_summator.sum(N)}")
cube_summator = CubeSummarator()
print(f"Сумма кубов 1..{N}: {cube_summator.sum(N)}")

print(f"\nПроверка по формулам:")
print(f"Сумма: {N*(N+1)//2}")
print(f"Сумма квадратов: {N*(N+1)*(2*N+1)//6}")
print(f"Сумма кубов: {(N*(N+1)//2)**2}")

```
