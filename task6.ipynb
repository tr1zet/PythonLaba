Напишите класс Selector. Экземпляр этого класса при инициализации получает список чисел. Вызов метода get_odds возвращает нечётные числа из первоначального списка, вызов get_evens — чётные. Числа должны идти в том же порядке, в котором они были в изначальном списке.


```python
class Selector:
    def __init__(self, numbers):
        self.numbers = numbers
    
    def get_odds(self):
        return [n for n in self.numbers if n % 2 != 0]
    
    def get_evens(self):
        return [n for n in self.numbers if n % 2 == 0]


selector = Selector([11, 12, 13, 14, 15, 16, 17, 8, 19, 10])
print("Нечётные:", selector.get_odds())
print("Чётные:", selector.get_evens())

```

Реализуйте класс Polynomial для вычисления значений многочленов при заданных x, а также для сложения многочленов.

poly = Polynomial(coefficients) — создать новый многочлен с коэффициентами coefficients.  
coefficients[0] — свободный член, coefficients[1] — множитель при \(x^1\), coefficients[2] — множитель при \(x^2\), и так далее.

y = poly(x) — вычислить значение многочлена в точке x.

\[y = c_0 + c_1 x^1 + c_2 x^2 + \dots + c_n x^n\]

poly_sum = poly1 + poly2 — результат сложения двух многочленов должен быть экземпляром Polynomial и содержать коэффициенты, которые получаются в результате сложения коэффициентов poly1 и poly2 при одинаковых степенях x.

**Примечание:** модуль numpy в этой задаче использовать нельзя.




```python
class Polynomial:
    def __init__(self, coefficients):
        self.coefficients = coefficients
    
    def __call__(self, x):
        result = 0
        for power, coeff in enumerate(self.coefficients):
            result += coeff * (x ** power)
        return result
    
    def __add__(self, other):
        max_len = max(len(self.coefficients), len(other.coefficients))
        new_coeffs = [0] * max_len
        for i in range(len(self.coefficients)):
            new_coeffs[i] += self.coefficients[i]
        for i in range(len(other.coefficients)):
            new_coeffs[i] += other.coefficients[i]
        return Polynomial(new_coeffs)
    
    def __str__(self):
        return f"Polynomial({self.coefficients})"



poly = Polynomial([10, -1])
print(poly(0))
print(poly(1))
print(poly(2))

print()

poly1 = Polynomial([0, 0, 1])
print(poly1(-2))
print(poly1(-1))
print(poly1(0))
print(poly1(1))
print(poly1(2))

poly2 = Polynomial([0, 0, 2])
print(poly2(-2))
print(poly2(-1))
print(poly2(0))
print(poly2(1))
print(poly2(2))
print()

poly3 = poly1 + poly2
print(poly3(-2))
print(poly3(-1))
print(poly3(0))
print(poly3(1))
print(poly3(2))
print()
```

### №8 (1 балл)

Реализуйте класс Queue, моделирующий работу очереди (принцип FIFO). Очередь может быть инициализирована различным числом элементов.

q = Queue(1, 2, 3, 4, 5) – создаёт очередь [1 -> 2 -> 3 -> 4 -> 5]

Класс Queue реализует следующие методы:

- append(values) – добавляет несколько значений в конец очереди (как минимум одно).
- copy() – создаёт копию данной очереди.
- pop() – вытаскивает и возвращает первый элемент очереди, при этом этот элемент из очереди удаляется. Если очередь пуста, то возвращает None.
- extend(queue) – расширяет данную очередь другой.
- next() – возвращает новую очередь, начинающуюся со второго элемента текущей.

Операторы и встроенные функции:

- queue1 + queue2 – склейка очередей создаёт новую увеличенную очередь.
- queue1 += queue2 – расширяет первую очередь второй.
- queue1 == queue2 – проверяет очереди на равенство всех элементов.
- queue >> N – создаёт новую очередь без первых N элементов.
- str(queue) – приводит очередь к строке вида [q1 -> q2 -> ... -> qn].
- next(queue) – аналогичное действие методу next().

```python
class Queue:
    def __init__(self, *args):
        self.items = list(args)
    
    def append(self, *values):
        for value in values:
            self.items.append(value)
    
    def copy(self):
        new_queue = Queue()
        new_queue.items = self.items.copy()
        return new_queue
    
    def pop(self):
        if len(self.items) == 0:
            return None
        return self.items.pop(0)
    
    def extend(self, queue):
        self.items.extend(queue.items)
    
    def next(self):
        if len(self.items) <= 1:
            return Queue()
        return Queue(*self.items[1:])
    
    def __len__(self):
        return len(self.items)
    
    def __str__(self):
        if not self.items:
            return "[]"
        return "[" + " -> ".join(str(item) for item in self.items) + "]"
    
    def __add__(self, other):
        new_queue = self.copy()
        new_queue.extend(other)
        return new_queue
    
    def __iadd__(self, other):
        self.extend(other)
        return self
    
    def __eq__(self, other):
        if not isinstance(other, Queue):
            return False
        return self.items == other.items
    
    def __rshift__(self, n):
        if n >= len(self.items):
            return Queue()
        return Queue(*self.items[n:])

    def __iter__(self):
        return self

    def __next__(self):
        if len(self.items) == 0:
            raise StopIteration
        return self.pop()



q1 = Queue(1, 2, 3)
print(q1)

q1.append(4, 5)
print(q1)

qx = q1.copy()
print(qx.pop())
print(qx)

q2 = q1.copy()
print(q2)

print(q1 == q2, id(q1) == id(q2))

q3 = q2.next()
print(q1)
print(q2)
print(q3)

print(q1 + q3)

q3.extend(Queue(1, 2))
print(q3)

q4 = Queue(1, 2)
q4 += q3 >> 4
print(q4)

q5 = next(q4)
print(q4)
print(q5)
```

### №13 (1 балл)

Реализуйте класс Weapon (name, damage, range) с методами hit(target) и __str__. Реализуйте класс BaseCharacter (pos_x, pos_y, hp) с методами move, is_alive, get_damage, get_coords. Реализуйте класс BaseEnemy, наследующий BaseCharacter, с методом hit(target) (только для MainHero) и __str__. Реализуйте класс MainHero, наследующий BaseCharacter, с методами hit (только для BaseEnemy), add_weapon, next_weapon, heal (макс. здоровье 200).


```python
import math

class Weapon:
    def __init__(self, name, damage, range_weapon):
        self.name = name
        self.damage = damage
        self.range = range_weapon
    
    def hit(self, actor, target):
        if not target.is_alive():
            print("Враг уже повержен")
            return
        ax, ay = actor.get_coords()
        tx, ty = target.get_coords()
        distance = math.sqrt((ax - tx) ** 2 + (ay - ty) ** 2)
        if distance > self.range:
            print(f"Враг слишком далеко для оружия '{self.name}'")
        else:
            print(f"Враг нанесен урон оружием '{self.name}' в размере {self.damage}")
            target.get_damage(self.damage)
    
    def __str__(self):
        return self.name


class BaseCharacter:
    def __init__(self, pos_x, pos_y, hp):
        self.pos_x = pos_x
        self.pos_y = pos_y
        self.hp = hp
        self.alive = True
    
    def move(self, delta_x, delta_y):
        self.pos_x += delta_x
        self.pos_y += delta_y
    
    def is_alive(self):
        return self.alive
    
    def get_damage(self, amount):
        self.hp -= amount
        if self.hp <= 0:
            self.alive = False
    
    def get_coords(self):
        return (self.pos_x, self.pos_y)


class BaseEnemy(BaseCharacter):
    def __init__(self, pos_x, pos_y, weapon, hp):
        super().__init__(pos_x, pos_y, hp)
        self.weapon = weapon
    
    def hit(self, target):
        if isinstance(target, MainHero):
            self.weapon.hit(self, target)
        else:
            print("Могу ударить только Главного героя")
    
    def __str__(self):
        return f"Враг на позиции ({self.pos_x}, {self.pos_y}) с оружием '{self.weapon}'"


class MainHero(BaseCharacter):
    def __init__(self, pos_x, pos_y, name, hp):
        super().__init__(pos_x, pos_y, hp)
        self.name = name
        self.weapons = []
        self.current_weapon_index = -1
        self.max_hp = 200
    
    def hit(self, target):
        if len(self.weapons) == 0:
            print("Я безоружен")
            return
        if isinstance(target, BaseEnemy):
            self.weapons[self.current_weapon_index].hit(self, target)
        else:
            print("Могу ударить только Врага")
    
    def add_weapon(self, weapon):
        if isinstance(weapon, Weapon):
            self.weapons.append(weapon)
            if len(self.weapons) == 1:
                self.current_weapon_index = 0
            print(f"Подобрал '{weapon}'")
        else:
            print("Это не оружие")
    
    def next_weapon(self):
        if len(self.weapons) == 0:
            print("Я безоружен")
        elif len(self.weapons) == 1:
            print("У меня только одно оружие")
        else:
            self.current_weapon_index = (self.current_weapon_index + 1) % len(self.weapons)
            print(f"Сменил оружие на '{self.weapons[self.current_weapon_index]}'")
    
    def heal(self, amount):
        self.hp += amount
        if self.hp > self.max_hp:
            self.hp = self.max_hp
        print(f"Попечился, теперь здоровья {self.hp}")


print("Создание персонажей ")
hero = MainHero(0, 0, "Герой ", 200)
enemy = BaseEnemy(3, 4, Weapon("Плевок", 20, 5), 50)

print(f"Герой: координаты {hero.get_coords()}, здоровье {hero.hp}")
print(f"Враг: {enemy}")

print("\nПроверка оружия ")
hero.hit(enemy)
hero.add_weapon(Weapon("Лук", 15, 10))
hero.hit(enemy)

hero.add_weapon(Weapon("Топор", 25, 15))
hero.next_weapon()
hero.next_weapon()

print("\nЛечение ")
hero.heal(30)
hero.heal(100)
```

### №14 (1 балл)

Реализуйте программу, имитирующую работу почтовых серверов и клиентов. Клиенты могут принимать и отправлять письма, которые хранятся на серверах.

- client = MailClient(server, user) — подключение к учётной записи.
- client.receive_mail() — получение почты (удаляется с сервера).
- client.send_mail(server1, user1, message) — отправка сообщения.

Программа должна позволять определять список серверов. При попытке отправить письмо на несуществующий сервер — сообщать об ошибке.

```python
class MailServer:
    def __init__(self, name):
        self.name = name
        self.mailboxes = {}
    
    def add_user(self, username):
        if username not in self.mailboxes:
            self.mailboxes[username] = []
            print(f"Пользователь {username} создан на сервере {self.name}")
    
    def send_mail(self, to_user, message):
        if to_user in self.mailboxes:
            self.mailboxes[to_user].append(message)
            return True
        return False
    
    def receive_mail(self, username):
        if username in self.mailboxes and self.mailboxes[username]:
            return self.mailboxes[username].pop(0)
        return None
    
    def has_mail(self, username):
        return username in self.mailboxes and len(self.mailboxes[username]) > 0


class MailClient:
    def __init__(self, server, user):
        self.server = server
        self.user = user
        server.add_user(user)
    
    def receive_mail(self):
        if self.server.has_mail(self.user):
            message = self.server.receive_mail(self.user)
            print(f"{self.user} получил письмо: {message}")
            return message
        print(f"{self.user}: Нет новых писем")
        return None
    
    def send_mail(self, target_server, target_user, message):
        if target_server is None:
            print("Ошибка: сервер не существует")
            return False
        success = target_server.send_mail(target_user, message)
        if success:
            print(f"{self.user} отправил письмо '{message}' пользователю {target_user} на сервер {target_server.name}")
        else:
            print(f"Ошибка: пользователь {target_user} не найден на сервере {target_server.name}")
        return success


class MailSystem:
    def __init__(self):
        self.servers = {}
    
    def create_server(self, name):
        if name not in self.servers:
            self.servers[name] = MailServer(name)
            print(f"Создан сервер {name}")
        return self.servers[name]
    
    def get_server(self, name):
        return self.servers.get(name)
    
    def list_servers(self):
        return list(self.servers.keys())


print("Cоздание почтовой системы")
system = MailSystem()

server1 = system.create_server("mail.ru")
server2 = system.create_server("gmail.com")

print(f"Список серверов: {system.list_servers()}")

print("\nCоздание клиентов")
daniil = MailClient(server1, "daniil")
anton = MailClient(server2, "anton")

print("\n Отправка писем ")
daniil.send_mail(server2, "anton", "Привет, Антон!")
anton.receive_mail()

print("\nПопытка отправить на несуществующий сервер ")
daniil.send_mail(system.get_server("yandex.ru"), "anton", "Тест")

print("\nОтправка с одного сервера ")
ivan = MailClient(server1, "ivan")
daniil.send_mail(server1, "ivan", "Привет,  Ваня")
ivan.receive_mail()
```

Программа преобразует список чисел в соответствии с командами: make_negative (только для положительных, умножает на -1), square (для всех, возводит в квадрат), strange_command (только для делящихся на 5, увеличивает на 1). Спроектировать так, чтобы добавление новых преобразований требовало минимальных изменений.

```python
class Command:
    def __init__(self, name, condition, transform):
        self.name = name
        self.condition = condition
        self.transform = transform
    
    def apply(self, numbers):
        result = []
        for n in numbers:
            if self.condition(n):
                result.append(self.transform(n))
            else:
                result.append(n)
        return result


def make_negative_command():
    return Command("make_negative", lambda x: x > 0, lambda x: -x)


def square_command():
    return Command("square", lambda x: True, lambda x: x ** 2)


def strange_command():
    return Command("strange_command", lambda x: x % 5 == 0, lambda x: x + 1)


def process_numbers(numbers, command_name, commands):
    for cmd in commands:
        if cmd.name == command_name:
            return cmd.apply(numbers)
    return numbers


numbers1 = [1, 0, -2, 30, -4]
result1 = process_numbers(numbers1, "make_negative", [
    make_negative_command(),
    square_command(),
    strange_command()
])
print(f"Вход: {numbers1}")
print(f"Команда: make_negative")
print(f"Выход: {result1}")

numbers2 = [1, 5, -2, 0, 30, -4]
result2 = process_numbers(numbers2, "square", [
    make_negative_command(),
    square_command(),
    strange_command()
])
print(f"Вход: {numbers2}")
print(f"Команда: square")
print(f"Выход: {result2}")

numbers3 = [1, 5, -2, 0, 30, -4]
result3 = process_numbers(numbers3, "strange_command", [
    make_negative_command(),
    square_command(),
    strange_command()
])
print(f"Вход: {numbers3}")
print(f"Команда: strange_command")
print(f"Выход: {result3}")
```

### №16 (2 балла)

Программа считывает определения математических функций в формате:
define f1 x * x
define f2 -1 * x
define f3 f1 + f2 Изначально существуют функции x и sqrt_fun. Выражение calculate f3 -5 -2 0 2 5` вычисляет функцию в указанных точках.

```python
import math

class Function:
    def __init__(self, name):
        self.name = name
    
    def evaluate(self, x):
        return x


class SqrtFunction(Function):
    def __init__(self):
        super().__init__("sqrt_fun")
    
    def evaluate(self, x):
        return math.sqrt(x)


class ConstantFunction(Function):
    def __init__(self, value):
        super().__init__(str(value))
        self.value = value
    
    def evaluate(self, x):
        return self.value


class BinaryOperation(Function):
    def __init__(self, name, left, op, right):
        super().__init__(name)
        self.left = left
        self.op = op
        self.right = right
    
    def evaluate(self, x):
        if isinstance(self.left, Function):
            left_val = self.left.evaluate(x)
        else:
            left_val = self.left
        
        if isinstance(self.right, Function):
            right_val = self.right.evaluate(x)
        else:
            right_val = self.right
        
        if self.op == '+':
            return left_val + right_val
        elif self.op == '-':
            return left_val - right_val
        elif self.op == '*':
            return left_val * right_val
        elif self.op == '/':
            return left_val / right_val


class FunctionCalculator:
    def __init__(self):
        self.functions = {
            "x": Function("x"),
            "sqrt_fun": SqrtFunction()
        }
    
    def define_function(self, name, left, op, right):
        left_obj = self._parse_operand(left)
        right_obj = self._parse_operand(right)
        self.functions[name] = BinaryOperation(name, left_obj, op, right_obj)
        print(f"Определена функция {name} = {left} {op} {right}")
    
    def _parse_operand(self, operand):
        if operand.isdigit() or (operand[0] == '-' and operand[1:].isdigit()):
            return int(operand)
        elif '.' in operand:
            return float(operand)
        else:
            return self.functions.get(operand, Function(operand))
    
    def calculate(self, func_name, points):
        func = self.functions.get(func_name)
        if not func:
            print(f"Функция {func_name} не найдена")
            return []
        
        results = []
        for point in points:
            results.append(func.evaluate(point))
        return results


print("Определение функций ")
calculator = FunctionCalculator()

calculator.define_function("f1", "x", "*", "x")
calculator.define_function("f2", "-1", "*", "x")
calculator.define_function("f3", "f1", "+", "f2")

print("\nВычисление значений ")
points = [-5, -2, 0, 2, 5]
results = calculator.calculate("f3", points)
print(f"f3 в точках {points}: {results}")

print("\nПроверка sqrt_fun ")
sqrt_results = calculator.calculate("sqrt_fun", [1, 4, 9, 16])
print(f"sqrt_fun в точках [1,4,9,16]: {sqrt_results}")

```

### №17 (2 балла)

Программа для планирования выступлений на конференциях: создание доклада (тема, время начала, длительность), контроль неперекрывания, предупреждения, суммарное время, самый продолжительный перерыв. Реализовать классы «доклад» и «конференция».


```python
from datetime import datetime, timedelta

class Talk:
    def __init__(self, title, start_time, duration_minutes):
        self.title = title
        self.start_time = start_time
        self.end_time = start_time + timedelta(minutes=duration_minutes)
        self.duration = duration_minutes
    
    def __str__(self):
        return f"'{self.title}' с {self.start_time.strftime('%H:%M')} до {self.end_time.strftime('%H:%M')}"
    
    def overlaps_with(self, other):
        return not (self.end_time <= other.start_time or self.start_time >= other.end_time)


class Conference:
    def __init__(self, name):
        self.name = name
        self.talks = []
    
    def add_talk(self, title, start_time_str, duration_minutes):
        start_time = datetime.strptime(start_time_str, "%H:%M")
        
        new_talk = Talk(title, start_time, duration_minutes)
        
        # Проверка на пересечение
        for talk in self.talks:
            if new_talk.overlaps_with(talk):
                print(f"Ошибка! Доклад '{title}' пересекается с докладом {talk}")
                return False
        
        self.talks.append(new_talk)
        self.talks.sort(key=lambda t: t.start_time)
        print(f"Доклад '{title}' добавлен")
        return True
    
    def remove_talk(self, title):
        for talk in self.talks:
            if talk.title == title:
                self.talks.remove(talk)
                print(f"Доклад '{title}' удален")
                return True
        print(f"Доклад '{title}' не найден")
        return False
    
    def total_talk_time(self):
        total = sum(talk.duration for talk in self.talks)
        print(f"Суммарное время всех докладов: {total} минут")
        return total
    
    def longest_break(self):
        if len(self.talks) < 2:
            print("Недостаточно докладов для определения перерыва")
            return 0
        
        max_break = 0
        for i in range(len(self.talks) - 1):
            break_start = self.talks[i].end_time
            break_end = self.talks[i + 1].start_time
            break_duration = (break_end - break_start).total_seconds() / 60
            if break_duration > max_break:
                max_break = break_duration
        
        print(f"Самый продолжительный перерыв: {max_break} минут")
        return max_break
    
    def show_schedule(self):
        if not self.talks:
            print("Нет докладов")
            return
        
        print(f"\nРасписание конференции '{self.name}' ")
        for i, talk in enumerate(self.talks, 1):
            print(f"{i}. {talk}")
        print()



print("Создание конференции ")
conf = Conference("IT Конференция 2024")

print("\n Добавление докладов ")
conf.add_talk("Вступительное слово", "10:00", 30)
conf.add_talk("Python для начинающих", "10:45", 60)
conf.add_talk("Машинное обучение", "12:00", 45)
conf.add_talk("Обед", "13:00", 60)

print("\n Попытка добавить пересекающийся доклад ")
conf.add_talk("Базы данных", "11:30", 60)

print("\n Показ расписания ")
conf.show_schedule()

print("\n Статистика ")
conf.total_talk_time()
conf.longest_break()

print("\nУдаление доклада ")
conf.remove_talk("Обед")
conf.show_schedule()
conf.longest_break()


```

### №18 (4 балла)

Программа для имитации работы файловой системы. Команды:
1. Создать директорию (поддерживаются вложенные пути, например "ааа/bbb/ccc").
2. Вывести список файлов и поддиректорий.
3. Записать строку в файл (при первой записи создаётся, при последующих перезаписывается).
4. Прочитать содержимое файла.

Директории должны хранить свои поддиректории и файлы, как в настоящих ФС.


```python
class File:
    def __init__(self, name, content=""):
        self.name = name
        self.content = content
    
    def write(self, content):
        self.content = content
    
    def read(self):
        return self.content
    
    def __str__(self):
        return self.name


class Directory:
    def __init__(self, name):
        self.name = name
        self.subdirectories = {}
        self.files = {}
    
    def create_directory(self, path):
        if not path:
            return False
        
        parts = path.split('/')
        current = self
        
        for i, part in enumerate(parts[:-1]):
            if part not in current.subdirectories:
                current.subdirectories[part] = Directory(part)
            current = current.subdirectories[part]
        
        dir_name = parts[-1]
        if dir_name not in current.subdirectories:
            current.subdirectories[dir_name] = Directory(dir_name)
            print(f"Создана директория: {path}")
            return True
        else:
            print(f"Директория {path} уже существует")
            return False
    
    def list_contents(self, path=""):
        if not path:
            current = self
        else:
            current = self._navigate_to(path)
            if current is None:
                print(f"Директория {path} не найдена")
                return
        
        print(f"\nСодержимое директории '{path if path else 'root'}':")
        print("Директории:", list(current.subdirectories.keys()) if current.subdirectories else "нет")
        print("Файлы:", list(current.files.keys()) if current.files else "нет")
    
    def write_file(self, path, content):
        if '/' in path:
            dir_path, file_name = path.rsplit('/', 1)
            parent = self._navigate_to(dir_path)
            if parent is None:
                print(f"Путь {dir_path} не найден")
                return False
        else:
            parent = self
            file_name = path
        
        if file_name in parent.files:
            parent.files[file_name].write(content)
            print(f"Файл {path} перезаписан")
        else:
            parent.files[file_name] = File(file_name, content)
            print(f"Файл {path} создан")
        return True
    
    def read_file(self, path):
        if '/' in path:
            dir_path, file_name = path.rsplit('/', 1)
            parent = self._navigate_to(dir_path)
            if parent is None:
                print(f"Путь {dir_path} не найден")
                return None
        else:
            parent = self
            file_name = path
        
        if file_name in parent.files:
            content = parent.files[file_name].read()
            print(f"Содержимое файла {path}: {content}")
            return content
        else:
            print(f"Файл {path} не найден")
            return None
    
    def _navigate_to(self, path):
        parts = path.split('/')
        current = self
        
        for part in parts:
            if part in current.subdirectories:
                current = current.subdirectories[part]
            else:
                return None
        return current


class FileSystem:
    def __init__(self):
        self.root = Directory("root")
    
    def create_directory(self, path):
        return self.root.create_directory(path)
    
    def list_contents(self, path=""):
        self.root.list_contents(path)
    
    def write_file(self, path, content):
        return self.root.write_file(path, content)
    
    def read_file(self, path):
        return self.root.read_file(path)
    
    def run_interactive(self):
        print("\n" + "="*50)
        print("Файловая система - интерактивный режим")
        print("="*50)
        print("Команды:")
        print("  mkdir <path>     - создать директорию")
        print("  ls [path]        - показать содержимое (по умолчанию root)")
        print("  write <path>     - записать в файл (будет запрошено содержимое)")
        print("  read <path>      - прочитать файл")
        print("  exit             - выход")
        print("="*50)
        
        while True:
            try:
                cmd = input("\n> ").strip()
                if not cmd:
                    continue
                
                parts = cmd.split(maxsplit=1)
                command = parts[0].lower()
                
                if command == "exit":
                    print("До свидания!")
                    break
                
                elif command == "mkdir":
                    if len(parts) < 2:
                        print("Ошибка: укажите путь")
                        continue
                    self.create_directory(parts[1])
                
                elif command == "ls":
                    path = parts[1] if len(parts) > 1 else ""
                    self.list_contents(path)
                
                elif command == "write":
                    if len(parts) < 2:
                        print("Ошибка: укажите путь к файлу")
                        continue
                    content = input("Введите содержимое файла: ")
                    self.write_file(parts[1], content)
                
                elif command == "read":
                    if len(parts) < 2:
                        print("Ошибка: укажите путь к файлу")
                        continue
                    self.read_file(parts[1])
                
                else:
                    print(f"Неизвестная команда: {command}")
            
            except KeyboardInterrupt:
                print("\nДо свидания!")
                break
            except EOFError:
                print("\nДо свидания!")
                break




fs = FileSystem()

print("1. Создание директорий:")
fs.create_directory("aaa")
fs.create_directory("aaa/bbb")
fs.create_directory("aaa/ccc")
fs.create_directory("aaa/bbb/ccc")

print("\n2. Просмотр содержимого:")
fs.list_contents()
fs.list_contents("aaa")
fs.list_contents("aaa/bbb")

print("\n3. Запись файлов:")
fs.write_file("1.txt", "Hello World")
fs.write_file("aaa/1.txt", "Hello from aaa")
fs.write_file("aaa/bbb/1.txt", "Hello from aaa/bbb")
fs.write_file("aaa/bbb/ccc/1.txt", "Hello from deep")

print("\n4. Перезапись файла:")
fs.write_file("1.txt", "Updated content")

print("\n5. Чтение файлов:")
fs.read_file("1.txt")
fs.read_file("aaa/1.txt")
fs.read_file("aaa/bbb/1.txt")
fs.read_file("aaa/bbb/ccc/1.txt")

print("\n6. Просмотр содержимого после создания файлов:")
fs.list_contents()
fs.list_contents("aaa")
fs.list_contents("aaa/bbb")
fs.list_contents("aaa/bbb/ccc")

print("\n7. Попытка прочитать несуществующий файл:")
fs.read_file("nonexistent.txt")

print("\n8. Попытка создать директорию, которая уже существует:")
fs.create_directory("aaa")

print("\n" + "="*50)
print("ЗАПУСК ИНТЕРАКТИВНОГО РЕЖИМА:")
print("="*50)

fs.run_interactive()
```
