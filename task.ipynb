Задание 6 Трехзначное число считается красивым, если в нём полусумма максимальной и
минимальной по значению цифр равна оставшейся цифре числа. Выведите для красивого
числа фразу "Вы ввели красивое число", а для остальных - "Жаль, вы ввели обычное число".

```python
num = int(input("Введите трехзначное число: "))
a = num // 100
b = (num // 10) % 10
c = num % 10
max_digit = max(a, b, c)
min_digit = min(a, b, c)
middle = (max_digit + min_digit) / 2
if middle == (a + b + c - max_digit - min_digit):
    print("Вы ввели красивое число")
else:
    print("Жаль, вы ввели обычное число")
```

Задание 8 Напишите программу, которая считывает рост претендентов в отряд космонавтов до тех
пор, пока не будет введен «!». А затем выводит на первой строчке количество подходящих
кандидатур, а на второй строке – минимальный и максимальный рост участников,
отобранных в новый отряд космонавтов. Рост космонавта не может быть больше 190 см и
меньше 150 см

```python
count = 0
min_height = 190
max_height = 150
while True:
    height = input("Введите рост (или ! для завершения): ")
    if height == "!":
        break
    height = int(height)
    if 150 <= height <= 190:
        count += 1
        if height < min_height:
            min_height = height
        if height > max_height:
            max_height = height
print(count)
print(min_height, max_height)
```

Задание 9 Напишите программу, которая имитирует проверку пароля, придуманного пользователем.
Пользователь вводит пару слов: пароль, а потом ещё раз его же, для подтверждения.
Если введённая пара не удовлетворяет одному из перечисленных ниже условий,
пользователь вводит пару паролей ещё раз, и так до тех пор, пока не будут выполнены все
условия (т. е. пока программа не выведет «OK»).
1) если первый пароль из пары, который ввёл пользователь короче 8 символов,
программа выводит на экран слово "Короткий!" и заново считывает пару словпаролей;
2) если же первый пароль из пары достаточно длинный, но в нём содержится сочетание
символов "123", программа выводит на экран слово "Простой!" и снова читает пару
слов-паролей;
3) если же предыдущие проверки пройдены успешно, но введённые слова-пароли не
совпадают, то программа выводит на экран слово "Различаются." и опять же читает
пару слов-паролей;
4) если же и третья проверка пройдена успешно, программа выводит "OK" (латинскими
буквами) и заканчивает свою работу.

```python
while True:
    p1 = input("Введите пароль: ")
    p2 = input("Повторите пароль: ")
    
    if len(p1) < 8:
        print("Короткий!")
        continue
    
    if "123" in p1:
        print("Простой!")
        continue
    
    if p1 != p2:
        print("Различаются.")
        continue
    
    print("OK")
    break
```

Задание 10 Напишите калькулятор для целочисленных вычислений, который обрабатывает серию
команд и выводит результат вычислений по каждой из них.
Каждая команда записывается так: сначала вводится первое число, затем символ операции,
затем — при необходимости — второе число. Символы операций (гарантируется, что
используются только эти символы):
1) «+» (сложить два числа),
2) «-» (вычесть из первого числа второе),
3) «*» (перемножить два числа),
4) «/» (поделить первое число на второе нацело),
5) «%» (получить остаток от деления первого числа на второе),
6) «!» (посчитать факториал от первого числа, ввод второго числа не производится),
7) «x» (вывести первое число и закончить выполнение программы, ввод второго числа
не производится).
Если пользователь попытается взять факториал от отрицательного числа или сделать ещё
что-то незаконное, то никакой результат для этой команды выводить не следует.

```python
def factorial(n):
    if n < 0:
        return None
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

while True:
    try:
        num1 = int(input("Введите первое число: "))
        op = input("Введите операцию (+, -, *, /, %, !, x): ")
        
        if op == "!":
            res = factorial(num1)
            if res is not None:
                print(res)
            continue
        elif op == "x":
            print(num1)
            break
        else:
            num2 = int(input("Введите второе число: "))
            
            if op == "+":
                print(num1 + num2)
            elif op == "-":
                print(num1 - num2)
            elif op == "*":
                print(num1 * num2)
            elif op == "/":
                if num2 != 0:
                    print(num1 // num2)
            elif op == "%":
                if num2 != 0:
                    print(num1 % num2)
    except:
        continue
```

Задание 13 Напишите программу построения прямоугольника n х m, состоящего из символов symb.
Фигура должна быть пустой, а не заполненной. То есть она должна состоять только из
контура.

```python
height = int(input("Введите высоту: "))
width = int(input("Введите ширину: "))
symb = input("Введите символ: ")
for i in range(height):
    for j in range(width):
        if i == 0 or i == height-1 or j == 0 or j == width-1:
            print(symb, end="")
        else:
            print(" ", end="")
    print()
```
