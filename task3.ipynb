Задание 4
Задача «Встречалось ли число раньше»

Условие
Во входной строке записана последовательность чисел через пробел. Для каждого числа выведите слово YES (в отдельной строке), если это число ранее встречалось в последовательности или NO, если не встречалось.

```python
print("Введите последовательность чисел через пробел:")
numbers = list(map(int, input().split()))
print("Cписок", numbers)
seen = set()
for num in numbers:
    if num in seen:
        print("YES")
    else:
        print("NO")
        seen.add(num)
```

Задание 5
Задача «Количество слов в тексте»

Условие
Дан текст: в первой строке записано число строк, далее идут сами строки. Определите, сколько различных слов содержится в этом тексте.

Словом считается последовательность непробельных символов идущих подряд, слова разделены одним или большим числом пробелов или символами конца строки.



```python
print("Введите количество строк:")
n = int(input())
print(f"Вы ввели количество строк: {n}")
print()

words = set()
for i in range(n):
    line = input(f"Введите строку {i + 1}: ")
    print(f"Вы ввели: {line}")
    for word in line.split():
        words.add(word)
    print()

print(f"Количество различных слов: {len(words)}")
```

Задание 6 Начальник кадровой службы хочет узнать, сколько мужчин-однофамильцев работает в организации. У него есть список фамилий, и на основании этого списка нужно вычислить количество фамилий, которые совпадают с другими. В первой строке указывается количество мужчин - сотрудников организации (N). Затем идут N строк с фамилиями этих сотрудников в произвольном порядке.
Выводится количество однофамильцев в организации.

```python
print("Введите количество сотрудников:")
n = int(input())
print(f"Вы ввели количество сотрудников: {n}")
print()

surnames = {}
for i in range(n):
    print(f"Введите фамилию сотрудника {i + 1}:")
    name = input().strip()
    print(f"Вы ввели фамилию: {name}")
    surnames[name] = surnames.get(name, 0) + 1
    print(f"Текущий словарь фамилий: {surnames}")
    print()

count = 0
for surname, freq in surnames.items():
    if freq > 1:
        print(f"Фамилия '{surname}' встречается {freq} раза")
        count += freq

print(f"Итог: Количество однофамильцев: {count}")
```

Задание 7 Задача «Номер появления слова»

Условие
В единственной строке записан текст. Для каждого слова из данного текста подсчитайте, сколько раз оно встречалось в этом тексте ранее.

Словом считается последовательность непробельных символов идущих подряд, слова разделены одним или большим числом пробелов или символами конца строки.

```python
print("Введите текст:")
text = input()
print(f"Вы ввели текст: {text}")
print()

words = text.split()
print(f"Разбиваем на слова: {words}")
print()

word_count = {}
print("Результаты подсчета:")
for word in words:
    if word in word_count:
        print(f"Слово '{word}' встречалось ранее {word_count[word]} раз(а)")
    else:
        print(f"Слово '{word}' встречается впервые")
    
    word_count[word] = word_count.get(word, 0) + 1
print()

print("Итоговый счетчик слов:")
for word, count in word_count.items():
    print(f"'{word}': {count}")
```

Задание 8 Задача «Словарь синонимов»

Условие
Вам дан словарь, состоящий из пар слов. Каждое слово является синонимом к парному ему слову. Все слова в словаре различны.

Для слова из словаря, записанного в последней строке, определите его синоним.

```python
print("Введите количество пар синонимов:")
n = int(input())
print(f"Вы ввели количество пар: {n}")
print()

synonym_dict = {}
for i in range(n):
    print(f"Введите пару синонимов {i + 1} (через пробел):")
    word1, word2 = input().split()
    print(f"Вы ввели: {word1} - {word2}")
    synonym_dict[word1] = word2
    synonym_dict[word2] = word1
    print(f"Текущий словарь: {synonym_dict}")
    print()

print("Словарь синонимов составлен:")
for word, synonym in synonym_dict.items():
    print(f"'{word}' - '{synonym}'")
print()

print("Введите слово для поиска синонима:")
word = input().strip()
print(f"Вы ищете синоним для слова: '{word}'")
print()

if word in synonym_dict:
    print(f"Результат: Синоним слова '{word}' - '{synonym_dict[word]}'")
else:
    print("Данного слова нет в словаре")
```

Задание 9 Задача «Выборы в США»

Условие
Как известно, в США президент выбирается не прямым голосованием, а путем двухуровневого голосования. Сначала проводятся выборы в каждом штате и определяется победитель выборов в данном штате. Затем проводятся государственные выборы: на этих выборах каждый штат имеет определенное число голосов — число выборщиков от этого штата. На практике, все выборщики от штата голосуют в соответствии с результами голосования внутри штата, то есть на заключительной стадии выборов в голосовании участвуют штаты, имеющие различное число голосов.

В первой строке дано количество записей. Далее, каждая запись содержит фамилию кандидата и число голосов, отданных за него в одном из штатов. Подведите итоги выборов: для каждого из участника голосования определите число отданных за него голосов. Участников нужно выводить в алфавитном порядке.

```python
print("Введите количество записей:")
n = int(input())
print(f"Вы ввели количество записей: {n}")
print()

votes = {}
for i in range(n):
    print(f"Введите запись {i + 1} (кандидат и количество голосов):")
    candidate, voice = input().split()
    voice = int(voice)
    print(f"Вы ввели: кандидат '{candidate}', голосов: {voice}")
    
    votes[candidate] = votes.get(candidate, 0) + voice
    print(f"Текущие результаты: {votes}")
    print()

print("Итоговые результа выборов (в алфавитном порядке):")
print("-" * 40)
for candidate in sorted(votes.keys()):
    print(f"{candidate}: {votes[candidate]} голосов")
print("-" * 40)
```

Задание 10
Задача «Права доступа»

Условие
В файловую систему одного суперкомпьютера проник вирус, который сломал контроль за правами доступа к файлам. Для каждого файла известно, с какими действиями можно к нему обращаться:

запись W,

чтение R,

запуск X.

В первой строке содержится число N — количество файлов содержащихся в данной файловой системе. В следующих N строчках содержатся имена файлов и допустимых с ними операций, разделенные пробелами. Далее указано число M — количество запросов к файлам. В последних M строках указан запрос вида Операция Файл. К одному и тому же файлу может быть применено любое количество запросов.

Вам требуется восстановить контроль над правами доступа к файлам (ваша программа для каждого запроса должна будет возвращать OK если над файлом выполняется допустимая операция, или же Access denied, если операция недопустима.

```python
operation_map = {
    "чтение": "R", 
    "запись": "W", 
    "запуск": "X",
    "read": "R",
    "write": "W", 
    "execute": "X"
}

print("Введите количество файлов:")
n = int(input())
print(f"Вы ввели количество файлов: {n}")
print()

files = {}
for i in range(n):
    print(f"Введите информацию о файле {i + 1} (имя и права через пробел):")
    data = input().split()
    filename = data[0]
    permissions = set(data[1:])
    files[filename] = permissions
    print(f"Файл: {filename}")
    print(f"Права доступа: {permissions}")
    print()

print("СПИСОК ФАЙЛОВ И ИХ ПРАВ:")
for filename, perms in files.items():
    print(f"{filename}: {perms}")
print()

print("Введите количество запросов:")
m = int(input())
print(f"Вы ввели количество запросов: {m}")
print()

print("Введите запросы (Операция Файл):")
for i in range(m):
    operation, filename = input().split()
    print(f"Запрос {i + 1}: операция '{operation}' к файлу '{filename}'")
    if operation in operation_map:
        op_symbol = operation_map[operation]
        print(f"Операция '{operation}' соответствует символу '{op_symbol}'")
    else:
        op_symbol = operation
        print(f"Операция '{operation}' не распознана, использую как есть: '{op_symbol}'")
    
    if filename in files:
        print(f"Файл '{filename}' найден")
        print(f"Его права доступа: {files[filename]}")
        if op_symbol in files[filename]:
            print("Результат: OK")
        else:
            print("Результат: Access denied")
    else:
        print(f"Результат: Access denied (файл '{filename}' не найден)")
    print()
```
