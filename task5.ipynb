### №1 (2 балла)

При помощи функций filter и map выполните задания:

1. Выведите все элементы списка, которые меньше 5.
2. Выведите все элементы списка, разделенные на два.
3. Выведите результат деления на два всех элементов списка, которые больше 17.
4. Напишите программу, которая подсчитает и выведет сумму квадратов всех двузначных чисел, делящихся на 9. При решении задачи используйте комбинацию функций filter, map, sum. Обратите внимание: на 9 должно делиться исходное двузначное число, а не его квадрат.

```python
# 1. Выведите все элементы списка, которые меньше 5
def task1_1(lst):
    return list(filter(lambda x: x < 5, lst))

# 2. Выведите все элементы списка, разделенные на два
def task1_2(lst):
    return list(map(lambda x: x / 2, lst))

# 3. Выведите результат деления на два всех элементов списка, которые больше 17
def task1_3(lst):
    return list(map(lambda x: x / 2, filter(lambda x: x > 17, lst)))

# 4. Сумма квадратов всех двузначных чисел, делящихся на 9
def task1_4():
    numbers = list(range(10, 100))
    return sum(map(lambda x: x**2, filter(lambda x: x % 9 == 0, numbers)))


test_list = [1, 3, 4, 7, 8, 12, 18, 20, 25, 30, 35, 37, 39, 41, 45, 50]
print("1. Элементы меньше 5:", task1_1(test_list))
print("2. Элементы, разделенные на 2:", task1_2(test_list))
print("3. Элементы >17, разделенные на 2:", task1_3(test_list))
print("4. Сумма квадратов двузначных чисел, кратных 9:", task1_4())
```

### №2 (1 балл)

Напишите функцию генератор factorials(n), генерирующую последовательность факториалов натуральных чисел.

```python
def factorials(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
        yield fact


for f in factorials(7):
    print(f, end=' ')
print()
```

### №3 (1 балл)

Напишите функцию генератор square_fibonacci(n), генерирующую последовательность квадратов чисел Фибоначчи.


```python
def square_fibonacci(n):
    a, b = 0, 1
    for i in range(n):
        a, b = b, a + b
        yield a * a

for f in square_fibonacci(7):
    print(f, end=' ')
print()
```

### №4 (1 балл)

```python
def russian_letters():
    for i in range(ord('а'), ord('я') + 1):
        yield chr(i)

count = 0
for letter in russian_letters():
    print(letter, end=' ')
    count += 1
    if count % 10 == 0:
        print()
print()
```

### №7 (1 балл)

Напишите функцию same_by(characteristic, objects), которая проверяет, все ли объекты имеют одинаковое значение некоторой характеристики, и возвращает True, если это так. Если значение характеристики для разных объектов отличается – то False. Для пустого набора объектов, функция должна возвращать True. Аргумент characteristic – это функция, которая принимает объект и вычисляет его характеристику.

```python
def same_by(characteristic, objects):
    if not objects:
        return True
    first = characteristic(objects[0])
    for obj in objects[1:]:
        if characteristic(obj) != first:
            return False
    return True

values1 = [0, 2, 10, 6]
values2 = [1, 2, 3, 4]

if same_by(lambda x: x % 2, values1):
    print('same')
else:
    print('different')

if same_by(lambda x: x % 2, values2):
    print('same')
else:
    print('different')
```

Напишите функцию print_operation_table(operation, num_rows=9, num_columns=9), которая принимает в качестве аргумента функцию, реализующую бинарную операцию вычисления элемента по номеру строки и столбца.
```python
def print_operation_table(operation, num_rows=9, num_columns=9):
    for i in range(1, num_rows + 1):
        row = []
        for j in range(1, num_columns + 1):
            row.append(str(operation(i, j)))
        print(' '.join(row))


print_operation_table(lambda x, y: x * y)
print('')
print_operation_table(lambda x, y: x * y, 5)
```

Напишите функцию ask_password(login, password, success, failure), которая получает логин и пароль пользователя и проверяет их правильность. Пароль считается правильным, если в нём содержится ровно три английские гласные буквы (гласными считать буквы a,e,i,o,u,y) и ровно такой же набор согласных (все буквы, кроме перечисленных шести), как в логине. Порядок и количество согласных также должно совпадать. Логин и пароль приводятся к нижнему регистру и передаются.

success и failure – коллбэки. Если пароль правильный, функция должна вызвать коллбэк success, передав ему в качестве аргумента логин. А если пароль был неверный - failure, передав ему в качестве аргументов логин и сообщение об ошибке (в таком порядке). Сообщение об ошибке должно быть одним из трёх вариантов:
- “Wrong number of vowels”, если в пароле неверное число гласных;
- “Wrong consonants”, если в пароле набор согласных отличается от набора согласных логина;
- “Everything is wrong”, если оба условия нарушены.

Также напишите функцию main(login, password), которая вызывает написанную функцию ask_password так, чтобы в случае успеха она печатала “Привет, {логин}!”, а в случае ошибки – “Кто-то пытался притвориться пользователем {логин}, но в пароле допустил ошибку: {текст ошибки, большими буквами}.”.

```python
def ask_password(login, password, success, failure):
    login = login.lower()
    password = password.lower()
    vowels = set('aeiouy') 
    #Подсчет гласных 
    password_vowels = sum(1 for c in password if c in vowels)
     #Получение согласных 
    login_consonants = [c for c in login if c.isalpha() and c not in vowels]
    password_consonants = [c for c in password if c.isalpha() and c not in vowels]
    # Проверка ошибок
    vowel_error = password_vowels != 3
    consonant_error = login_consonants != password_consonants
    
    if vowel_error and consonant_error:
        failure(login, "Everything is wrong")
    elif vowel_error:
        failure(login, "Wrong number of vowels")
    elif consonant_error:
        failure(login, "Wrong consonants")
    else:
        success(login)

def main(login, password):
    def success(login):
        print(f"Привет, {login}!")
    
    def failure(login, error):
        print(f"Кто-то пытался притвориться пользователем {login}, но в пароле допустил ошибку: {error.upper()}.")
    
    ask_password(login, password, success, failure)

main("anastasia", "nysatos")
main("eugene", "aangi")
ask_password("anastasia", "nysatos", 
             lambda login: print('super'), 
             lambda login, err: print('bad'))
```

Напишите программу для обработки текста. На вход вашей программы подаётся многострочный текст, причём число строк заранее неизвестно. Ваша задача – пронумеровать слова в нём, начиная с нуля, а затем вывести только те слова, которые начинаются с большой буквы. Перед словом необходимо вывести номер первого вхождения этого слова в текст. Слова необходимо отсортировать в лексикографическом порядке. Для решения этой задачи вам помогут итераторы sys.stdin и enumerate.

```python
import sys

def task14():
    words_dict = {}
    first_occurrence = {}
    
    test_lines = [
        "Ехал Грека через реку.",
        "Видит грека в реке рак.", 
        "Cунул в реку руку Грека",
        "Рак за руку Греку цап"
    ]
    
    for line in test_lines:
        line_words = line.strip().split()
        for word in line_words:
            if word not in words_dict:
                words_dict[word] = 1
                first_occurrence[word] = len(words_dict) - 1
            else:
                words_dict[word] += 1
    
    # Выбираем слова с большой буквы
    capital_words = [word for word in words_dict.keys() if word and word[0].isupper()]
    capital_words.sort(key=lambda x: x.lower())
    
    print("\nРезультат (номер первого вхождения и слово):")
    for word in capital_words:
        print(first_occurrence[word], word)

task14()
```

### №18 (3 балла)

Напишите генератор декораторов check_password, т. е. функцию, которая возвращает декоратор. Генератор декораторов принимает в качестве параметра пароль, и получившийся декоратор должен закрыть функцию этим паролем. Декоратор будет применяться следующим образом:

```python
def check_password_factory(password):
    def decorator(func):
        def wrapper(*args, **kwargs):
            user_password = input("Введите пароль для доступа к калькулятору: ")
            if user_password == password:
                return func(*args, **kwargs)
            else:
                print("Неверный пароль! Доступ запрещен.")
                return None
        return wrapper
    return decorator

@check_password_factory("daniil")
def calculator(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        return a / b if b != 0 else "Ошибка: деление на ноль"


print("Калькулятор")
result = calculator(2, 3, "+")
if result:
    print(f"Результат: {result}")
```

### №19 (3 балла)
Напишите декоратор cached, который будет кэшировать результат вызова функции. Пример того, как можно будет использовать ваш декоратор:

```python
def cached(func):
    cache = {}
    
    def wrapper(*args):
        if args in cache:
            print(f"Берем из кэша для аргументов {args}")
            return cache[args]
        print(f"Вычисляем для аргументов {args}")
        result = func(*args)
        cache[args] = result
        return result
    
    return wrapper

@cached
def fib(n):
    if n <= 2:
        return 1
    return fib(n - 1) + fib(n - 2)


print("fib(10) =", fib(10))
print("fib(10) =", fib(10))  
print("fib(5) =", fib(5))

```
